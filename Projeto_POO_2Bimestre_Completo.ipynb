## PARTE 1 – Classe Principal e Objetos

## PARTE 2 – Métodos e Herança

**Tarefas:**
- Criar pelo menos 2 métodos de ação para a classe principal.
- Criar uma subclasse herdando da classe principal.
- Adicionar pelo menos 1 novo atributo e 1 novo método exclusivo na subclasse.
- Utilizar `super()` no construtor da subclasse.

**Código:**

```python
# Desenvolva aqui o código da Parte 2
# Classe principal
class Planta:
    def __init__(self, nome, tipo, frequencia_rega):
        self.nome = nome
        self.tipo = tipo
        self.frequencia_rega = frequencia_rega  # em dias
        self.ultima_rega = 0  # dias desde a última rega

    # Método 1: Regar a planta
    def regar(self):
        self.ultima_rega = 0
        print(f"💧 {self.nome} foi regada.")

    # Método 2: Passagem de dias sem regar
    def passar_dia(self):
        self.ultima_rega += 1
        print(f"🌞 Passou 1 dia para {self.nome}. Já se passaram {self.ultima_rega} dias desde a última rega.")

    def exibir_dados(self):
        print(f"Nome: {self.nome}")
        print(f"Tipo: {self.tipo}")
        print(f"Frequência de rega: {self.frequencia_rega} dias")
        print(f"Dias desde a última rega: {self.ultima_rega}")
        print("-" * 30)

# Subclasse: PlantaInterior
class PlantaInterior(Planta):
    def __init__(self, nome, tipo, frequencia_rega, local_da_casa):
        super().__init__(nome, tipo, frequencia_rega)  # Chamada ao construtor da classe mãe
        self.local_da_casa = local_da_casa  # novo atributo

    # Novo método exclusivo da subclasse
    def verificar_luz(self):
        print(f"🔎 Verificando luz para {self.nome} no(a) {self.local_da_casa}. Luz indireta recomendada.")

    # Sobrescrevendo exibir_dados para incluir o local
    def exibir_dados(self):
        super().exibir_dados()
        print(f"Local da casa: {self.local_da_casa}")
        print("-" * 30)

# Criando objetos
rosa = Planta("Rosa", "Flor", 3)
samambaia = PlantaInterior("Samambaia", "Folhagem", 2, "sala de estar")

# Usando métodos
rosa.exibir_dados()
rosa.passar_dia()
rosa.regar()

samambaia.exibir_dados()
samambaia.passar_dia()
samambaia.verificar_luz()
```

## PARTE 3 – Encapsulamento e Abstração

**Tarefas:**
- Tornar pelo menos 1 atributo da classe principal privado (`__atributo`).
- Criar métodos `get` e `set` para o atributo privado.
- Implementar um método que contenha lógica interna (ex: cálculo, verificação).

**Código:**

```python
# Desenvolva aqui o código da Parte 3
from datetime import datetime, timedelta

class Planta:
    def __init__(self, nome, tipo, frequencia_rega):
        self.nome = nome
        self.tipo = tipo
        self.__frequencia_rega = frequencia_rega  # Atributo privado
        self.ultima_rega = datetime.now()  # Data da última rega

    # Getter
    def get_frequencia_rega(self):
        return self.__frequencia_rega

    # Setter
    def set_frequencia_rega(self, nova_frequencia):
        if nova_frequencia > 0:
            self.__frequencia_rega = nova_frequencia
            print(f"Frequência de rega da planta '{self.nome}' atualizada para {nova_frequencia} dias.")
        else:
            print("❌ Frequência inválida. Deve ser maior que zero.")

    # Método com lógica interna
    def precisa_regar(self):
        proxima_rega = self.ultima_rega + timedelta(days=self.__frequencia_rega)
        hoje = datetime.now()
        if hoje >= proxima_rega:
            print(f"💧 Atenção! A planta '{self.nome}' precisa ser regada hoje.")
        else:
            dias_restantes = (proxima_rega - hoje).days
            print(f"⏳ Ainda faltam {dias_restantes} dia(s) para regar a planta '{self.nome}'.")

    # Método de exibição
    def exibir_dados(self):
        print(f"🌿 Planta: {self.nome} | Tipo: {self.tipo} | Rega: a cada {self.__frequencia_rega} dias")


# === Testando a classe com encapsulamento ===

planta1 = Planta("Lavanda", "Aromática", 3)
planta1.exibir_dados()
planta1.precisa_regar()

# Alterando a frequência com o setter
planta1.set_frequencia_rega(5)

# Verificando getter
print("Nova frequência:", planta1.get_frequencia_rega())
```

## PARTE 4 – Polimorfismo e Banco de Dados

**Tarefas:**
- Implementar polimorfismo: sobrescrever um método na subclasse.
- Integrar o sistema com banco de dados (MySQL ou SQLite).
- Implementar inserção (INSERT), consulta (SELECT) e exclusão (DELETE) de registros no banco de dados.

**Código:**

```python
# Instale o conector MySQL se necessário
# !pip install mysql-connector-python

import mysql.connector
from mysql.connector import Error

# === CLASSES ===

class Planta:
    def __init__(self, nome, tipo, frequencia_rega):
        self.nome = nome
        self.tipo = tipo
        self.frequencia_rega = frequencia_rega

    def exibir_dados(self):
        print(f"🌱 Planta comum: {self.nome} | Tipo: {self.tipo} | Rega: a cada {self.frequencia_rega} dias")

class PlantaInterior(Planta):
    def __init__(self, nome, tipo, frequencia_rega, local_da_casa):
        super().__init__(nome, tipo, frequencia_rega)
        self.local_da_casa = local_da_casa

    def exibir_dados(self):
        print(f"🏠 Planta de Interior: {self.nome} | Tipo: {self.tipo} | Rega: a cada {self.frequencia_rega} dias | Local: {self.local_da_casa}")


# === CONEXÃO COM BANCO MySQL ===

def criar_conexao():
    try:
        conn = mysql.connector.connect(
            host='localhost',     # ajuste se seu MySQL estiver em outro host
            user='root',          # usuário padrão do XAMPP
            password='',          # senha do seu MySQL, provavelmente vazia no XAMPP padrão
            database='jardim_db'  # nome do banco que você criou
        )
        if conn.is_connected():
            print("✅ Conectado ao MySQL!")
            return conn
    except Error as e:
        print(f"Erro ao conectar: {e}")
        return None

conn = criar_conexao()
cursor = conn.cursor()

# === CRIAR TABELA (se não existir) ===
cursor.execute('''
    CREATE TABLE IF NOT EXISTS plantas (
        id INT AUTO_INCREMENT PRIMARY KEY,
        nome VARCHAR(100) NOT NULL,
        tipo VARCHAR(100) NOT NULL,
        frequencia_rega INT NOT NULL,
        local_da_casa VARCHAR(100)
    )
''')
conn.commit()

# === FUNÇÕES PARA INTERAÇÃO ===

def inserir_planta(planta):
    sql = '''
        INSERT INTO plantas (nome, tipo, frequencia_rega, local_da_casa)
        VALUES (%s, %s, %s, %s)
    '''
    valores = (planta.nome, planta.tipo, planta.frequencia_rega, getattr(planta, 'local_da_casa', None))
    cursor.execute(sql, valores)
    conn.commit()
    print(f"✅ Planta '{planta.nome}' inserida com sucesso.")

def listar_plantas():
    cursor.execute("SELECT * FROM plantas")
    registros = cursor.fetchall()
    print("\n📋 Plantas no banco de dados:")
    for p in registros:
        id, nome, tipo, freq, local = p
        if local:
            planta = PlantaInterior(nome, tipo, freq, local)
        else:
            planta = Planta(nome, tipo, freq)
        planta.exibir_dados()

def deletar_planta_por_nome(nome):
    cursor.execute("DELETE FROM plantas WHERE nome = %s", (nome,))
    conn.commit()
    print(f"🗑️ Planta '{nome}' removida do banco de dados.")

# === EXEMPLOS DE USO ===

# Criando objetos
planta1 = Planta("Rosa", "Flor", 3)
planta2 = PlantaInterior("Samambaia", "Folhagem", 2, "sala de estar")

# Inserindo plantas no banco
inserir_planta(planta1)
inserir_planta(planta2)

# Listando plantas
listar_plantas()

# Deletando planta pelo nome
deletar_planta_por_nome("Rosa")

# Listando novamente para confirmar exclusão
listar_plantas()

# Fechando conexão (quando terminar)
conn.close()
print("🔒 Conexão fechada.")
```

## PARTE 5 – Projeto Final e Organização

**Tarefas:**
- Consolidar todas as partes do projeto em um único script organizado.
- Comentar o código explicando cada parte.
- Testar todas as funcionalidades.

**Código:**

```python
import sqlite3

# ============================== teste
# CLASSES
# ==============================

# Classe principal: Planta
class Planta:
    def __init__(self, nome, tipo, frequencia_rega):
        self.nome = nome
        self.tipo = tipo
        self.frequencia_rega = frequencia_rega  # em dias

    def exibir_dados(self):
        print(f"🌱 Planta comum: {self.nome} | Tipo: {self.tipo} | Rega: a cada {self.frequencia_rega} dias")

# Subclasse: PlantaInterior
class PlantaInterior(Planta):
    def __init__(self, nome, tipo, frequencia_rega, local_da_casa):
        super().__init__(nome, tipo, frequencia_rega)  # herda do construtor da classe Planta
        self.local_da_casa = local_da_casa  # atributo exclusivo

    def verificar_luz(self):
        print(f"🔎 {self.nome} está no(a) {self.local_da_casa}. Verifique se há luz indireta.")

    # Polimorfismo: sobrescrevendo o método da classe mãe
    def exibir_dados(self):
        print(f"🏠 Planta de Interior: {self.nome} | Tipo: {self.tipo} | Rega: a cada {self.frequencia_rega} dias | Local: {self.local_da_casa}")

# ==============================
# BANCO DE DADOS (SQLite)
# ==============================

# Conexão com o banco de dados
conn = sqlite3.connect('jardim.db')
cursor = conn.cursor()

# Criação da tabela se não existir
cursor.execute('''
    CREATE TABLE IF NOT EXISTS plantas (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nome TEXT NOT NULL,
        tipo TEXT NOT NULL,
        frequencia_rega INTEGER NOT NULL,
        local_da_casa TEXT
    )
''')
conn.commit()

# Inserir planta no banco
def inserir_planta(planta):
    cursor.execute('''
        INSERT INTO plantas (nome, tipo, frequencia_rega, local_da_casa)
        VALUES (?, ?, ?, ?)
    ''', (planta.nome, planta.tipo, planta.frequencia_rega, getattr(planta, 'local_da_casa', None)))
    conn.commit()
    print("✅ Planta cadastrada com sucesso!")

# Listar todas as plantas do banco
def listar_plantas():
    print("\n📋 Plantas cadastradas:")
    cursor.execute("SELECT * FROM plantas")
    registros = cursor.fetchall()
    if not registros:
        print("Nenhuma planta encontrada.")
    for p in registros:
        id, nome, tipo, freq, local = p
        if local:
            planta = PlantaInterior(nome, tipo, freq, local)
        else:
            planta = Planta(nome, tipo, freq)
        planta.exibir_dados()

# Excluir planta pelo nome
def deletar_planta_por_nome(nome):
    cursor.execute("DELETE FROM plantas WHERE nome = ?", (nome,))
    conn.commit()
    print(f"🗑️ Planta '{nome}' removida com sucesso!")

# ==============================
# MENU INTERATIVO
# ==============================

def menu():
    while True:
        print("\n=== MENU DO JARDIM ===")
        print("1. Cadastrar planta comum")
        print("2. Cadastrar planta de interior")
        print("3. Listar todas as plantas")
        print("4. Deletar planta pelo nome")
        print("5. Sair")
        opcao = input("Escolha uma opção: ")

        if opcao == '1':
            nome = input("Nome da planta: ")
            tipo = input("Tipo da planta: ")
            frequencia = int(input("Frequência de rega (em dias): "))
            planta = Planta(nome, tipo, frequencia)
            inserir_planta(planta)

        elif opcao == '2':
            nome = input("Nome da planta: ")
            tipo = input("Tipo da planta: ")
            frequencia = int(input("Frequência de rega (em dias): "))
            local = input("Local da casa (ex: sala, cozinha): ")
            planta = PlantaInterior(nome, tipo, frequencia, local)
            inserir_planta(planta)

        elif opcao == '3':
            listar_plantas()

        elif opcao == '4':
            nome = input("Digite o nome da planta a ser excluída: ")
            deletar_planta_por_nome(nome)

        elif opcao == '5':
            print("🌼 Saindo do sistema. Até logo!")
            break

        else:
            print("Opção inválida. Tente novamente.")

# ==============================
# EXECUÇÃO DO SISTEMA
# ==============================

menu()
conn.close()  # Fecha a conexão com o banco
```
